```python
import sys
import pandas as pd
import os
import dotenv

sys.path.append('/Users/mlchrzan/Library/CloudStorage/OneDrive-Personal/Professional/Data Science/Personal Projects/pairadigm')
import pairadigm as pr

dotenv.load_dotenv()
```

```python
prior_sample = pd.read_csv('Data/annotation_sample.csv')
emobank = pd.read_csv('Data/emobank.csv')
```

```python
# Pull a random sample of 10 items from emobank that are NOT in prior_sample
sample = emobank[~emobank['id'].isin(prior_sample['id'])].sample(10, random_state=42).reset_index(drop=True)
sample
```

```python
paired_sample = pr.pair_items(sample['text'].tolist(), num_pairs_per_item=3, random_seed=42)
paired_sample.to_excel('Data/training_sample.xlsx', index=False)

paired_sample
```
